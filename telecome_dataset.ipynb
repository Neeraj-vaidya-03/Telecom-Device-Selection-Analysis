```python
df['Px'] = df['Px_h'] * df['Px_w']
df['Px']
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Px'], kde=True)
plt.axvline(df['Px'].mean(), color='red', linestyle='--', label=f"Mean: {df['Px'].mean():.2f}")
plt.axvline(df['Px'].median(), color='green', linestyle='-', label=f"Median: {df['Px'].median():.2f}")
plt.legend()
plt.show()

print("Mean:", df['Px'].mean())
print("Median:", df['Px'].median())
```

```python
con2 = df['Px'] >= df['Px'].median()
con2
```

```python
df[['Scr_h', 'Scr_w']].head()
```

```python
df['Scr_d'] = np.sqrt(df['Scr_h']**2 + df['Scr_w']**2)
df['Scr_d']
```

```python

plt.figure(figsize=(8, 5))
sns.histplot(df['Scr_d'], kde=True)
q25, q50, q75 = np.percentile(df['Scr_d'], [25, 50, 75])
plt.axvline(q25, color='orange', linestyle='--', label=f"Q1 (25%): {q25:.2f}")
plt.axvline(q50, color='green', linestyle='-', label=f"Q2/Median (50%): {q50:.2f}")
plt.axvline(q75, color='red', linestyle='--', label=f"Q3/Upper Quartile (75%): {q75:.2f}")
plt.legend()
plt.show()

print("Quartiles (25%, 50%, 75%):", q25, q50, q75)
```

```python
upper_quartile = df['Scr_d'].quantile(0.75)
con3 = df['Scr_d'] >= upper_quartile
con3
```
